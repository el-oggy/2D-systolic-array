# PYNQ-Z2 Wireless Hand Gesture Server

This notebook runs the **UDP Gesture Server** directly on the PYNQ-Z2 board.
It listens for real-time gesture commands from any laptop webcam running `full_screen_hand_tracker.py`.

### Actions:
- **Palm Open** &rarr; Green RGB LED (LD4)
- **Fist** &rarr; Red RGB LED (LD4)
- **Thumbs Up** &rarr; User LED 1 (LD1) ON (Latched)
- **Thumbs Down** &rarr; User LED 1 (LD1) OFF (Latched)

```python
# Set required environment variables
import os, sys
os.environ.setdefault('BOARD', 'Pynq-Z2')
os.environ.setdefault('XILINX_XRT', '/usr')

# Check PYNQ environment
try:
    from pynq import Overlay
    from pynq.lib import LED, RGBLED
    print("✅ PYNQ library found! Hardware LED control ready.")
except ImportError as e:
    print(f"⚠️ PYNQ import issue: {e}")
```
